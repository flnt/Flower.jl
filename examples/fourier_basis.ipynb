```python
import numpy as np
import pandas as pd
# import ace_tools as tools

# Sample points
x = np.linspace(-2, 2, 500)

# Set maximum power
N = 5

# Build design matrix for sin^n and cos^n
basis = []
names = []
for n in range(1, N + 1):
    basis.append(np.sin(2 * np.pi * x) ** n)
    names.append(f"sin^{n}")
for n in range(1, N + 1):
    basis.append(np.cos(2 * np.pi * x) ** n)
    names.append(f"cos^{n}")

A = np.column_stack(basis)
b = -0.5 * np.ones_like(x)

# Solve least squares
coeffs, *_ = np.linalg.lstsq(A, b, rcond=None)

# Prepare DataFrame
df = pd.DataFrame({
    "Basis Function": names,
    "Coefficient": coeffs
})

# tools.display_dataframe_to_user("Least‑Squares Coefficients for Powers of sin/cos", df)
print(df)
```
